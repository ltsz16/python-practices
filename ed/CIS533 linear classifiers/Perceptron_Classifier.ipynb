This function can assume that it is only called when an update should be performed. Hint: You simply need to implement $\mathbf{w} \leftarrow \mathbf{w} + y \mathbf{x}$.

```python
def perceptron_update(x, y, w):
    """
    function w=perceptron_update(x,y,w);

    Updates the perceptron weight vector w using x and y
    Input:
    x : input vector of d dimensions (d)
    y : corresponding label (-1 or +1)
    w : weight vector of d dimensions

    Output:
    w : weight vector after updating (d)
    """
    # YOUR CODE HERE
    w = w + 1 * x * y
    # END OF YOUR CODE

    return w


# Little test
y = -1
x = np.random.randint(low=-2, high=2, size=10)
w = np.random.randint(low=0, high=2, size=10) * 2 - 1
print(f"y     = {y}")
print(f"x     = {x}")
print(f"w_old = {w}")

w_new = perceptron_update(x, y, w)
print(f"w_new = {w_new}")
```

## Part Two: Implement Perceptron [Graded]

Implement function **`perceptron`**. This should contain a loop that calls **`perceptron_update`** until it converges or the maximum iteration count, 100, has been reached. Make sure you randomize the order of the training data on each iteration (you can use [**`np.random.permutation()`**](https://numpy.org/doc/stable/reference/random/generated/numpy.random.permutation.html) to do this.)

Note that the pseudocode below assumes that $b$ is absorbed into $\mathbf{w}$. In other words, the last dimension of $\mathbf{x}_i$ is 1 and the last dimension of $\mathbf{w}$ is the bias term. When implementing the **`perceptron`** function, make sure to take this into consideration and modify the data and weight vector accordingly.

Pseudocode:

<center>
    <img src="perceptron_pseudocode.png" width="700px" alt="Perceptron pseudocode" />
</center>

```python
def perceptron(xs, ys):
    """
    function w=perceptron(xs,ys);

    Returns the weight vector learned by the Perceptron classifier.

    Input:
        xs: n input vectors of d dimensions (nxd matrix)
        ys: n labels (-1 or +1)

    Output:
        w: weight vector (d)
        b: bias term
    """
    n, d = xs.shape  # So we have n input vectors with d dimensions
    w = np.zeros(d)
    b = 0.0

    # YOUR CODE HERE
    while True:
        cnt = 0
        for x, y in zip(xs,ys):
            resp = np.dot(w,x) + b
            if y * resp <=0:
                w = w + y * x
                b = b + y
                cnt += 1
        if cnt == 0:
            break
    # END OF YOUR CODE

    return (w, b)
```

## Part Three: Visualize Your Perceptron [Ungraded]

Once you have **`perceptron`** implemented, we can visualize how it finds `w` that classify points in a 2-dimensional plane. In this example, we center the data by subtracting the common mean vector of $\mathbf{x}_i$ from each of $\mathbf{x}_i$. Consequently, the perceptron algorithm will learn the bias term (the last dimension of $\mathbf{w}$ as 0).

The next cell defines a function `plot_perceptron_2D` that you will use to visualize your perceptron model.

```python
def plot_perceptron_2D(x, y_true, w_true, y_pred, w_pred, b):
    w_pred = w_pred.flatten()
    w_true = w_true.flatten()

    fig = plt.figure()
    plt.scatter(x[:, 0], x[:, 1], c=y_true, cmap="viridis")  # Plot points
    grid_lim = np.max(abs(x[:, :-1]))

    def calc_x2(b, x1, w):
        return -(b + w[0] * x1) / w[1]

    # Make arrow for w_true: normalize by magnitude, then stretch to grid lim
    w_true_mag = np.sqrt(np.sum(w_true[:-1] ** 2))
    plt.quiver(
        [0],
        [0],
        [(w_true[0] / w_true_mag) * grid_lim],
        [(w_true[1] / w_true_mag) * grid_lim],
        scale=1,
        units="xy",
        color="black",
    )
    plt.annotate(
        "w_true",
        xy=w_true[:-1] / w_true_mag * grid_lim + [0.05, 0.0],
        fontsize=12,
        color="black",
    )

    # Make decision line perpendicular to arrow
    plt.plot(
        [grid_lim, -grid_lim],
        [calc_x2(b, grid_lim, w_true), calc_x2(b, -grid_lim, w_true)],
        color="black",
        label="True decision line",
    )

    # Repeat for w_pred
    w_pred_mag = np.sqrt(np.sum(w_pred[:-1] ** 2))
    plt.quiver(
        [0],
        [0],
        [(w_pred[0] / w_pred_mag) * grid_lim],
        [(w_pred[1] / w_pred_mag) * grid_lim],
        scale=1,
        units="xy",
        color="red",
    )
    plt.annotate(
        "w_pred",
        xy=w_pred[0] / w_pred_mag * grid_lim + [0.0, 0.1],
        fontsize=12,
        color="red",
    )
    plt.plot(
        [grid_lim, -grid_lim],
        [calc_x2(b, grid_lim, w_pred), calc_x2(b, -grid_lim, w_pred)],
        color="red",
        label="Pred decision line",
    )

    plt.title(f"True and Predicted Decision Line")
    plt.xlim(-grid_lim, grid_lim)
    plt.ylim(-grid_lim, grid_lim)
    plt.legend(bbox_to_anchor=(1, 1), loc="upper right")
    #plt.grid()
    plt.show()
```

### Part Three: Make Predictions [Graded]

Implement **`classify_linear`** that applies the weight vector (with implicit bias term) to the input vector. Make sure that the predictions returned are either `1` or `-1`.

```python
def classify_linear(xs, w, b=None):
    """
    function preds=classify_linear(xs,w,b)

    Make predictions with a linear classifier
    
    Input:
        xs: n input vectors of d dimensions (nxd) [could also be a single vector of d dimensions]
        w: weight vector of dimensionality d
        b: bias (scalar)

    Output:
        preds: predictions (1xn)
    """
    w = w.flatten()
    predictions = np.zeros(xs.shape[0])

    # YOUR CODE HERE
    if b is None:
        b = 0
    w = w.flatten()
    resp = xs@w + b
    predictions = np.where(resp >=0,1,-1)
    # END OF YOUR CODE
    
    return predictions
```

Run the next cell to check that your linear classifier correctly classifies the points in a linearly separable dataset.

```python
def linear_test1():
    """
    Check that predictions are only -1 or 1
    """
    xs = np.random.rand(50000, 20) - 0.5  # Draw random data
    xs = np.hstack([xs, np.ones((50000, 1))])
    w0 = np.random.rand(20 + 1)
    w0[-1] = -0.1                         # Assign bias of -0.1
    ys = classify_linear(xs, w0)
    uniquepredictions = np.unique(ys)
    return set(uniquepredictions) == {-1, 1}


def linear_test2():
    """
    Check that all predictions are correct for linearly separable data
    """
    xs = np.random.rand(1000, 2) - 0.5  # Draw random data
    xs = np.hstack([xs, np.ones((1000, 1))])

    # Define a random hyperplane with bias -0.1
    w0 = np.array([0.5, -0.3, -0.1])
    
    # Assign labels according to this hyperplane (so you know it is linearly separable)
    ys = np.sign(xs.dot(w0))

    # Original hyperplane (w0, b0) should classify all ones correctly
    result = all(np.sign(ys * classify_linear(xs, w0)) == 1.0)
    
    return result


runtest(linear_test1, "linear_test1")
runtest(linear_test2, "linear_test2")
```

Scikit-learn provides a [perceptron model](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Perceptron.html). We provide a brief demonstration of its usage below.

```python
N, d = 500, 2
np.random.seed(1)
x = np.random.rand(N, d)

# Subtracting mean puts features in range [-0.5, 0.5]
x -= np.mean(x)

# Add constant dimension -- a column of 1s at the end
x = np.hstack([x, np.ones((N, 1))])

w_true = np.random.rand(d)
w_true -= np.mean(w_true)  # Only rotates weight vector in 2D space
w_true = np.append(w_true, [0.0])  # Add bias term equal to 0
y_true = classify_linear_grader(x, w_true).astype(int)

# Change label of a data point.
# This way perceptron will never learn w_true,
# and we can plot both w_true and w_pred distinctly.
y_true[np.random.randint(N)] *= -1
```

```python
# Initialize the Perceptron
clf = Perceptron(max_iter=100, fit_intercept=True)

# Train on all features except the constant feature that was manually added
clf.fit(x[:, :-1], y_true)
w_pred_sklearn = np.concatenate([clf.coef_.flatten(), clf.intercept_.flatten()])

# The function classify_linear_grader() expects x to have 1s as the last column
y_pred_sklearn = classify_linear_grader(x, w_pred_sklearn).astype(int)
```
